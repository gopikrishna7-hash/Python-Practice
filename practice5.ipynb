```python
# # You are given a number k and a list arr[] that contains integers.
# You need to return list of numbers that are less than k.

def lessThan( arr, k):
    res=[]
    for num in arr:
        if num < k:
            res.append(num)
    return res
                
                
lessThan([2,3,2,43,23],22)
```

```python
# Given an array arr[], return the floor value of the mean of its elements.

def findMean(arr):
    total_sum=0
    for num in arr:
        total_sum+=num
    avg=total_sum //len(arr)
    
    return avg
findMean([23,22,43,2,12])
            
```
