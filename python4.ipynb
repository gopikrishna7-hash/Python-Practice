# Comparision Operator (==,>=,<=,!=,<,>)

```python
y,z=23,12

print(y ==z)
print(y>=z)
print(y <=z)
print(y!=z)
print(y<z)
print(y>z)
```

# Logical Operators ( and, or, not)

```python
a,b=True, False

print(a and b)
print(a or b)
print(not b)
```

# Identity Operators (is, is not)

```python
x=[4,5,6]
y=[4,5,6]
c=x

print(x is y) # Reference / Memory Address is differnent
print(x is not y)
print(x is c)
```

# Membership Operator (in, not in)

```python
Veg=['brinjal', 'Tomato', 'Radish', 'Onion']

print('Tomato' in Veg)
print('Radish' in Veg)
print('Radish'  not in Veg)
```
