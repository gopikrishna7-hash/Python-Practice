```python
# # Given two strings txt and pat, return the 0-based index of the first occurrence of the substring pat in txt.
# If pat is not found, return -1.
# # Note: You are not allowed to use the inbuilt function.
def firstOccurence(txt,pat):
        #code here
        
        if not pat:
            return -1

        n = len(txt)
        m = len(pat)

        for i in range(n - m + 1):
            matched = True

            for j in range(m):
                if txt[i + j] != pat[j]:
                    matched = False
                    break

            if matched:
                return i

        return -1
txt="SameSame"
pat="e"
firstOccurence(txt,pat)
```

```python
# Given a decimal integer n, convert it and return its binary equivalent as a string.
def decToBinary( n):
        # code here
        res=""
        while n> 0:
            rem=n%2
            res=str(rem)+ res
            n=n//2
        
        return res
decToBinary(84)
```
