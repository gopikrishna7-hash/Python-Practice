```python
# Given an array arr[], find the size of the array and in O(1) Complexity

def getSize(arr):
    return len(arr)
getSize([3,4,2,2,4])
            
```

```python
# # Sum the List
# Given a list arr that contains integers, return the sum of the list.
def listSum(arr):
    #code here
    total_sum=0
    for i in arr:
        total_sum+=i
    return total_sum
listSum([234,244,23,45,78])
```

```python
# You are given three inputs a, b, c. You need to create a list and 
# append a, b, c to the list and then return that list.
def appendToList(a,b,c):
    arr=[]
    arr.append(a)
    arr.append(b)
    arr.append(c)
    return arr
    
    
```
